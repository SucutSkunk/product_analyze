## 과제 1. 평가 문항 기반 독립 과제

### 문제 3-1. 유입 경로별 고객사 이탈률 비교하기

#### 문제 설명

Ravenstack 마케팅팀은 고객사가 유입된 경로에 따라 이탈 수준이 다른지 확인하려고 합니다. 유입 경로별 고객사 수와 이탈률을 비교하여 우선 점검할 유입 경로를 찾으세요.

> 이 과제는 필수 2에서 수행한 그룹별 지표 집계와 해석을 새로운 컬럼에 동일하게 적용하는 문제입니다.

#### 요구사항

1. `accounts`를 `referral_source`별로 그룹화하세요.
2. 다음 값을 집계하여 `source_metrics`를 만드세요.
   - 전체 고객사 수
   - 이탈 고객사 수
   - 고객사 이탈률
3. 고객사 이탈률이 높은 순서로 정렬하세요.
4. 이탈률은 소수점 둘째 자리의 백분율로 출력하세요.
5. 이탈률이 가장 높은 유입 경로를 찾으세요.
6. 해당 유입 경로를 우선 점검 대상으로 정하고, 확인할 개선 방향을 한 가지 제안하세요.
7. 유입 경로별 이탈률을 좋은 지표의 세 조건으로 평가하세요.
   - 행동 가능성
   - 비교 가능성
   - 이해 용이성

#### 해석 질문

**Q1.** 고객사 이탈률이 가장 높은 유입 경로는 무엇인가요?  
**Q2.** 유입 경로별 이탈률은 마케팅팀의 행동으로 어떻게 연결할 수 있나요?  
**Q3.** 유입 경로별 이탈률은 좋은 지표의 세 조건을 충족하나요?
**Q4.** KPI·보조 지표·허무 지표는 어떻게 다른가요? 고객사 이탈을 줄이려는 목적에서 각각의 지표 예시와 선정 이유를 제시하세요. 허무 지표는 어떤 맥락에서 성과 판단에 도움이 되지 않는지도 설명하세요.

#### 제출 결과

- `source_metrics` 집계 코드와 결과
- 우선 점검할 유입 경로
- 개선 방향
- 좋은 지표의 조건 평가
- Q1~Q4 답변

```python
# 과제 1 코드를 작성하세요.
import pandas as pd

accounts = pd.read_csv('ravenstack_accounts.csv')
subscriptions = pd.read_csv('ravenstack_subscriptions.csv')
# 1~2. referral_source별 그룹화 및 집계
source_metrics = (
    accounts.groupby('referral_source')
    .agg(
        total_accounts=('account_id', 'nunique'),   # 전체 고객사 수
        churned_accounts=('churn_flag', 'sum'),     # 이탈 고객사 수
        churn_rate=('churn_flag', 'mean')           # 고객사 이탈률
    )
    # 3. 이탈률 높은 순 정렬
    .sort_values('churn_rate', ascending=False)
)

# 4. 이탈률을 소수점 둘째 자리 백분율로 출력
source_display = source_metrics.copy()
source_display['churn_rate'] = source_display['churn_rate'].map(lambda x: f'{x:.2%}')
print(source_display)

# 5. 이탈률이 가장 높은 유입 경로
top_source = source_metrics['churn_rate'].idxmax()
print(f"\n이탈률이 가장 높은 유입 경로: {top_source} "
      f"({source_metrics.loc[top_source, 'churn_rate']:.2%})")
print(f"전체 고객사 이탈률: {accounts['churn_flag'].mean():.2%}")
```

### 과제 1 답변 작성란

- **Q1.** 고객사 이탈률이 가장 높은 유입 경로는 **event(30.21%)**입니다. 96개사 중 29개사가 이탈했으며, 전체 고객사 이탈률(22.00%)보다 약 8%p 높고 가장 낮은 partner(14.61%)의 약 두 배 수준입니다.

- **Q2.** 전체 이탈률만 보면 이탈이 있다는 사실만 알 수 있지만, 유입 경로별로 나누면 "event 경로의 이탈이 특히 높다"처럼 문제가 발생한 위치를 좁힐 수 있습니다. 이를 바탕으로 마케팅팀은 이벤트 유입 고객사의 특성을 추가 분석하고, 이벤트 이후 온보딩을 보강하는 등 개선 행동을 정할 수 있습니다. 반대로 이탈률이 낮은 partner 경로는 유입 비중을 늘리는 방향을 검토할 수 있습니다.

- **Q3.** 세 조건을 충족한다고 볼 수 있습니다.
  - **행동 가능성:** 이탈률이 높은 경로를 찾으면 원인을 분석하고 우선 개선할 대상을 정할 수 있습니다.
  - **비교 가능성:** 모든 경로에 "이탈 고객사 수 ÷ 전체 고객사 수"라는 같은 기준을 적용하므로 경로끼리 비교할 수 있고, 같은 방식으로 지속적으로 측정해 변화도 비교할 수 있습니다.
  - **이해 용이성:** "이 경로로 들어온 고객사 중 몇 %가 떠났는가"라는 의미가 분명해 마케팅팀도 쉽게 이해할 수 있습니다.
  - 단, 경로별 고객사 수가 89~114개사로 많지 않아 소수 고객사의 이탈에도 이탈률이 크게 변할 수 있다는 점은 함께 고려해야 합니다.

- **Q4.**
  - **KPI**는 핵심 목표의 달성 여부를 판단하는 지표입니다("목표를 달성하고 있는가?"). 고객사 이탈을 줄이려는 목적에서는 **전체 고객사 이탈률**을 KPI로 볼 수 있습니다. 이 값이 낮아지면 목표가 달성되고 있음을 직접 보여주기 때문입니다.
  - **보조 지표**는 KPI의 변화와 원인을 이해하는 데 도움을 주는 지표입니다("KPI가 왜 변했는가?"). **유입 경로별 고객사 이탈률**이 예시입니다. 전체 이탈률이 변했을 때 어느 경로에서 변화가 생겼는지 보여주어 원인 분석과 개선 대상 선정에 활용할 수 있습니다.
  - **허무 지표**는 숫자는 좋아 보이지만 목표나 행동과의 연결이 약한 지표입니다("이 숫자가 좋아진 것이 실제로 무엇을 의미하는가?"). **누적 가입 고객사 수**가 예시입니다. 누적 값은 고객사가 이탈해도 줄지 않고 계속 증가하기 때문에, 이탈을 줄이려는 맥락에서는 현재 고객사가 떠나고 있는지나 어디에 문제가 있는지 알려주지 못합니다. 이 숫자만 보고 성과가 좋다고 판단하면 실제 이탈 문제를 놓칠 수 있습니다.
